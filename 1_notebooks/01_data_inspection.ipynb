# Data Inspection

## Introduction
This notebook is an initial inspection of the UCI Bank Marketing dataset. The goal is to understand the structure, data types, missing values, and basic characteristics of the raw data before the exploratory analysis.

This step also helps to identify potential issues such as inconsistent categories, outliers, or imbalance that will be addressed later during preprocessing.

```python
#import the required libraries
import pandas as pd
```

```python
#Read in csv
df = pd.read_csv('../01_data/bank-full.csv', sep = ';')
```

```python
#Preview data
df.head()
```

```python
#Check df shape - i.e. number of rows and columns
print(df.shape)
```

```python
#Check column names
print(df.columns)
```

```python
#Check data types
print(df.dtypes)
```

```python
#Check distributions
df.describe()
```

```python
#Check for missing values
df.isna().sum()
```

```python
#Check for empty strings
(df == '').any()
```

```python
#Check for duplicated rows
df.duplicated().sum()
```

```python
#Check unique variables in categorical columns
#job, marital, education, default, housing, loan, contact, month, poutcome
for col in df.select_dtypes(include = 'object'):
    print('')
    print(f"Column: {col}")
    print(df[col].value_counts())
```

## Summary

The dataset contains 45,211 rows and 17 columns with no missing values and no empty strings. Several categorical columns contain an "unknown" category, which might require cleaning later. Numeric columns such as balance and duration show wide ranges that needs to be explored further during EDA. The target variable is imbalanced, with more "no" responses than "yes", which will need to be addressed during preprocessing.
